```python
import pandas as pd
import mysql.connector
from sqlalchemy import create_engine
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid", context="notebook")

# Professional color palettes
colors = sns.color_palette("Set2")
bright = sns.color_palette("Spectral")
blue_palette = sns.color_palette("Blues_d")
green_palette = sns.color_palette("Greens_d")
purple_palette = sns.color_palette("Purples_d")

plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["axes.titleweight"] = "bold"
plt.rcParams["axes.titlesize"] = 16
plt.rcParams["axes.labelsize"] = 11
```

```python
conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="17March#3",
    database="dataco_supply"
)

print("MySQL Connected Successfully!")
```

```python
query = "SELECT * FROM supply_chain"

df = pd.read_sql(query, conn)

df.head()
```

```python
df['order date (DateOrders)'] = pd.to_datetime(
    df['order date (DateOrders)'],
    errors='coerce'
)

df['shipping date (DateOrders)'] = pd.to_datetime(
    df['shipping date (DateOrders)'],
    errors='coerce'
)
```

```python
df[['order date (DateOrders)', 'shipping date (DateOrders)']].dtypes
```

```python
df['Delivery_Days'] = (
    df['shipping date (DateOrders)'] -
    df['order date (DateOrders)']
).dt.days

df['Delivery_Days'].describe()
```

```python
total_sales = df['Sales'].sum()
print("Total Sales:", round(total_sales, 2))
```

```python
total_profit = df['Order Profit Per Order'].sum()
print("Total Profit:", round(total_profit, 2))
```

```python
total_orders = df['Order Id'].nunique()
print("Total Orders:", total_orders)
```

```python
total_customers = df['Customer Id'].nunique()
print("Total Customers:", total_customers)
```

```python
aov = total_sales / total_orders
print("Average Order Value:", round(aov, 2))
```

```python
monthly_sales = (
    df.groupby(df['order date (DateOrders)'].dt.to_period('M'))['Sales']
    .sum()
)

monthly_sales.plot(
    figsize=(12,5),
    marker='o',
    title='Monthly Sales Trend'
);
```

```python
market_sales = (
    df.groupby('Market')['Sales']
    .sum()
    .sort_values(ascending=False)
)

market_sales
```

```python
market_sales.plot(
    kind='bar',
    figsize=(10,5),
    title='Sales by Market'
);
```

```python
country_sales = (
    df.groupby('Order Country')['Sales']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

country_sales.plot(
    kind='bar',
    figsize=(10,5),
    title='Top 10 Countries by Sales'
);
```

```python
top_products = (
    df.groupby('Product Name')['Sales']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_products
```

```python
top_products.plot(
    kind='barh',
    figsize=(10,6),
    title='Top 10 Products by Sales'
);
```

```python
category_analysis = (
    df.groupby('Category Name')
    .agg(
        Sales=('Sales','sum'),
        Profit=('Order Profit Per Order','sum'),
        Orders=('Order Id','nunique')
    )
    .sort_values('Sales', ascending=False)
)

category_analysis
```

```python
category_analysis.sort_values(
    'Profit',
    ascending=False
).head(10)
```

```python
category_analysis['Profit'].sort_values(
    ascending=False
).head(10).plot(
    kind='bar',
    figsize=(10,5),
    title='Top Categories by Profit'
);
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))

plt.scatter(
    df['Sales'],
    df['Order Profit Per Order'],
    alpha=0.3
)

plt.xlabel('Sales')
plt.ylabel('Profit')
plt.title('Sales vs Profit')

plt.show()
```

```python
monthly_sales = (
    df.groupby(df['order date (DateOrders)'].dt.to_period('M'))['Sales']
    .sum()
    .reset_index()
)

monthly_sales['Month'] = monthly_sales[
    'order date (DateOrders)'
].astype(str)

plt.figure(figsize=(14,6))

sns.lineplot(
    data=monthly_sales,
    x='Month',
    y='Sales',
    marker='o',
    linewidth=3,
    color='#2E86DE'
)

plt.fill_between(
    range(len(monthly_sales)),
    monthly_sales['Sales'],
    alpha=0.15,
    color='#2E86DE'
)

plt.title("Monthly Sales Trend", fontsize=18, fontweight='bold')
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
top_products = (
    df.groupby('Product Name')['Sales']
    .sum()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)

plt.figure(figsize=(12,7))

sns.barplot(
    data=top_products,
    x='Sales',
    y='Product Name',
    hue='Product Name',
    palette='magma',
    legend=False
)

plt.title("Top 10 Products by Sales", fontsize=18, fontweight='bold')
plt.xlabel("Sales")
plt.ylabel("Product")

plt.tight_layout()
plt.show()
```

```python
category_sales = (
    df.groupby('Category Name')['Sales']
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

plt.figure(figsize=(12,6))

sns.barplot(
    data=category_sales,
    x='Sales',
    y='Category Name',
    hue='Category Name',
    palette='Spectral',
    legend=False
)

plt.title("Sales by Product Category", fontsize=18, fontweight='bold')
plt.xlabel("Total Sales")
plt.ylabel("Category")

plt.tight_layout()
plt.show()
```

```python
discount_analysis = (
    df.groupby('Category Name')
    .agg(
        Avg_Discount=('Order Item Discount', 'mean'),
        Total_Profit=('Order Profit Per Order', 'sum'),
        Total_Sales=('Sales', 'sum')
    )
    .reset_index()
)

plt.figure(figsize=(12,7))

sns.scatterplot(
    data=discount_analysis,
    x='Avg_Discount',
    y='Total_Profit',
    size='Total_Sales',
    hue='Total_Sales',
    palette='viridis',
    sizes=(100,800),
    alpha=0.8
)

for _, row in discount_analysis.iterrows():
    plt.text(
        row['Avg_Discount'],
        row['Total_Profit'],
        row['Category Name'],
        fontsize=8
    )

plt.title(
    "Discount vs Profit by Category",
    fontsize=18,
    fontweight='bold'
)

plt.xlabel("Average Discount")
plt.ylabel("Total Profit")

plt.tight_layout()
plt.show()
```
